```python
import ROOT
```

```python
file=ROOT.TFile("gaussrand.root", "RECREATE")
tree=ROOT.TTree("Tree","Gaussian distributed random numbers")
```

```python
x=0
N=1000
```

```python
tree.Branch("x",x,"x/D")

for i in range(0,N):
    x=ROOT.TRandom3(0).Gaus(0,1)
    tree.Fill()

tree.Write()
file.Close()

print("TFile generated containing",N,"Gaussian distributed random numbers")
```
